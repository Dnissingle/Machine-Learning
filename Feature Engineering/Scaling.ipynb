```python
import pandas as pd
import numpy as np
```

```python
# 1. Absolute Maximum Scaling

df = pd.read_csv ('housing.csv')
df = df.select_dtypes(include = np.number)
df.head()
```

```python
max_abs = np.max(np.abs(df), axis=0)

scaled_df = df / max_abs

scaled_df.head()
```

```python
# 2. Min-Max Scaling
from sklearn.preprocessing import MinMaxScaler

scaler = MinMaxScaler()
scaled_data = scaler.fit_transform(df)
scaled_df = pd.DataFrame(scaled_data, columns=df.columns)

scaled_df.head()
```

```python
df.head()
```

```python
# 3. Normalization (Vector Normalization) used in classification ans clustering

from sklearn.preprocessing import Normalizer
import pandas as pd

# Initialize the vector normalizer
scaler = Normalizer()

# Corrected the typo 'fit_transfrom' to 'fit_transform'
scaled_data = scaler.fit_transform(df)

# Convert the resulting numpy array back into a Pandas DataFrame
scaled_df = pd.DataFrame(scaled_data, columns=df.columns)
scaled_df.head()
```

```python
# 4. Standardization

from sklearn.preprocessing import StandardScaler

scaler = StandardScaler()
scaled_data = scaler.fit_transform(df)
scaled_df = pd.DataFrame(scaled_data,
                         columns=df.columns)
print(scaled_df.head())
```

```python
# 5. Robust Scaling
from sklearn.preprocessing import RobustScaler

scaler = RobustScaler()
scaled_data = scaler.fit_transform(df)
scaled_df = pd.DataFrame(scaled_data,
                         columns=df.columns)
print(scaled_df.head())
```
