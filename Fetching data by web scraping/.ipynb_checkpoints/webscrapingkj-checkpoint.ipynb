```python
jobs_listing = soup.find_all('div', class_='job-section p-2')
```

```python
len(jobs_listing)
```

```python
import pandas as pd

company = []
job = []
salary = []
location = []
jobLevel = []
workingTime = []
experience = []

# Helper function to safely extract text and avoid AttributeError
def get_text(element, default="N/A"):
    return element.text.strip() if element else default

for i in jobs_listing:
    # 1. Safely extract single elements
    company.append(get_text(i.find('p', class_='pe-2')))
    job.append(get_text(i.find('h2')))
    salary.append(get_text(i.find('span', class_='salary fw-semibold')))
    location.append(get_text(i.find('span', class_='district-text')))

    # 2. Safely extract multiple elements sharing the same class
    time_cards = i.find_all('div', class_='job-card-time')
    
    # Check if we found enough elements before using indices [0] and [1]
    level_val = time_cards[0].text.strip() if len(time_cards) > 0 else "N/A"
    time_val = time_cards[1].text.strip() if len(time_cards) > 1 else "N/A"
    exp_val = time_cards[2].text.strip() if len(time_cards) > 2 else "N/A" # Note: Adjusted to index 2 assuming experience is the 3rd card

    jobLevel.append(level_val)
    workingTime.append(time_val)
    experience.append(exp_val)

# Create DataFrame
d = {
    'Company': company, 
    'Job': job,
    'Salary': salary, 
    'Level': jobLevel,
    'Work Time': workingTime, 
    'Experience': experience
}
df = pd.DataFrame(d)
df
```
