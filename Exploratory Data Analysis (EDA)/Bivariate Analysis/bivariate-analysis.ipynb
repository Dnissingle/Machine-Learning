```python
# 1. Scatterplot

import seaborn as sns
import matplotlib.pyplot as plt
import pandas as pd

# Load the dataset
df = pd.read_csv('housingprice.csv')

# 1. Broad matrix overview across major numerical drivers
numerical_cols = ['Price', 'Area', 'Build Area', 'Road Width', 'Bedroom']
sns.pairplot(df[numerical_cols + ['City']], hue='City', palette='Set2', diag_kind='kde')
plt.suptitle('Multivariate Distribution Across Cities', y=1.02)
plt.show()
```

```python
import seaborn as sns
import matplotlib.pyplot as plt

plt.figure(figsize=(10, 6))
sns.barplot(
    data=df, 
    x='Bedroom', 
    y='Floors', 
    hue='Bedroom',      # Keeps separate colors for each bar
    palette='muted',
    errorbar=None,
    legend=False        # Suppresses the redundant legend since X-axis already tells us the bedrooms
)

# Fixed: Labels match the actual plotted data variables
plt.title('Average Number of Floors by Bedroom Count')
plt.xlabel('Number of Bedrooms')
plt.ylabel('Average Floors')
plt.tight_layout()
plt.show()
```
