```python
import pandas as pd
import seaborn as sns
```

```python
df = pd.read_csv('fertility.csv')
```

```python
# 1. Categorical DATA
# a. Countnplot

sns.countplot(df['Childish diseases'])
#df['Diagnosis'].value_counts().plot(kind='bar')
```
